```python
import numpy as np

# 1. Input Image (5x5 matrix jahan numbers pixels ko represent karte hain)
#
image = np.array([
    [3, 1, 0, 2, 1],
    [2, 0, 1, 3, 0],
    [4, 1, 0, 1, 2],
    [1, 3, 2, 0, 1],
    [0, 2, 1, 4, 2]
])

# 2. Convolution Layer
# Kernel 3x3 slide karta hai aur multiply kar ke sum karta hai[cite: 1]
kernel = np.array([
    [1,  0, -1],
    [1,  0, -1],
    [1,  0, -1]
])

# Feature map ka size 3x3 hoga (stride 1, no padding)[cite: 1]
feature_map = np.zeros((3, 3))
for i in range(3):
    for j in range(3):
        feature_map[i, j] = np.sum(image[i:i+3, j:j+3] * kernel)

print("2. Convolution Output (Feature Map):\n", feature_map)

# 3. ReLU Activation
# Formula: f(x) = max(0, x) - Negative numbers 0 ho jate hain[cite: 1]
relu_output = np.maximum(0, feature_map)
print("\n3. ReLU Output:\n", relu_output)

# 4. Pooling Layer (Max Pooling 2x2)
# Har 2x2 window sirf sab se bada number rakhti hai[cite: 1]
pooled_output = np.zeros((2, 2))
for i in range(2):
    for j in range(2):
        pooled_output[i, j] = np.max(relu_output[i:i+2, j:j+2])

print("\n4. Max Pooling Output (2x2):\n", pooled_output)

# 5. Flatten + Fully Connected
# 2D grid ko 1D vector mein badal diya jata hai[cite: 1]
flattened = pooled_output.flatten()
print("\n5. Flattened Vector:\n", flattened)

# Class Weights (Cat, Dog, Bird)[cite: 1]
weights_cat = np.array([0.3, 0.1, 0.2, 0.1])
weights_dog = np.array([0.1, 0.2, 0.1, 0.3])
weights_bird = np.array([0.0, 0.4, 0.1, 0.2])

# Logits calculate karna (Har neuron input ko weight se multiply kar ke sum karta hai)[cite: 1]
logit_cat = np.dot(flattened, weights_cat)
logit_dog = np.dot(flattened, weights_dog)
logit_bird = np.dot(flattened, weights_bird)

logits = np.array([logit_cat, logit_dog, logit_bird])
print("\nFully Connected Logits (Cat, Dog, Bird):", logits)

# 6. Softmax Output
# Logits ko probabilities mein convert karna jo ≈ 100% ka sum banati hain[cite: 1]
exp_logits = np.exp(logits)
softmax_probs = exp_logits / np.sum(exp_logits)

print("\n6. Softmax Probabilities:")
print(f"Cat:  {softmax_probs[0]*100:.0f}%") # 81% aana chahiye[cite: 1]
print(f"Dog:  {softmax_probs[1]*100:.0f}%") # 13% aana chahiye[cite: 1]
print(f"Bird: {softmax_probs[2]*100:.0f}%") # 5% aana chahiye[cite: 1]
```

```python
import numpy as np

image = np.random.randint(0, 6, size=(5, 5))
print("1. Random Input Image:\n", image)

kernel = np.array([
    [1,  0, -1],
    [1,  0, -1],
    [1,  0, -1]
])

feature_map = np.zeros((3, 3))
for i in range(3):
    for j in range(3):
        feature_map[i, j] = np.sum(image[i:i+3, j:j+3] * kernel)

print("\n2. Convolution Output (Feature Map):\n", feature_map)

relu_output = np.maximum(0, feature_map)
print("\n3. ReLU Output:\n", relu_output)

pooled_output = np.zeros((2, 2))
for i in range(2):
    for j in range(2):
        pooled_output[i, j] = np.max(relu_output[i:i+2, j:j+2])

print("\n4. Max Pooling Output (2x2):\n", pooled_output)

flattened = pooled_output.flatten()
print("\n5. Flattened Vector:\n", flattened)

weights_cat = np.array([0.3, 0.1, 0.2, 0.1])
weights_dog = np.array([0.1, 0.2, 0.1, 0.3])
weights_bird = np.array([0.0, 0.4, 0.1, 0.2])

logit_cat = np.dot(flattened, weights_cat)
logit_dog = np.dot(flattened, weights_dog)
logit_bird = np.dot(flattened, weights_bird)

logits = np.array([logit_cat, logit_dog, logit_bird])

logits_safe = logits - np.max(logits)
exp_logits = np.exp(logits_safe)
softmax_probs = exp_logits / np.sum(exp_logits)

print("\n6. Softmax Probabilities:")
print(f"Cat:  {softmax_probs[0]*100:.2f}%")
print(f"Dog:  {softmax_probs[1]*100:.2f}%")
print(f"Bird: {softmax_probs[2]*100:.2f}%")
```
